```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../data/processed/daily_consumption.csv")

data["Date"] = pd.to_datetime(data["Date"])

data = data.sort_values("Date").reset_index(drop=True)

data.head()
```

```python
data["Year"] = data["Date"].dt.year

data["Month"] = data["Date"].dt.month

data["Day"] = data["Date"].dt.day

data["Day_of_Week"] = data["Date"].dt.dayofweek

data["Is_Weekend"] = data["Day_of_Week"].isin([5, 6]).astype(int)
```

```python
data[[
    "Date",
    "Year",
    "Month",
    "Day",
    "Day_of_Week",
    "Is_Weekend"
]].head(10)
```

```python
data["Consumption_Lag_1"] = data["Daily_Consumption_kWh"].shift(1)
```

```python
data["Consumption_Lag_7"] = data["Daily_Consumption_kWh"].shift(7)
```

```python
data["Consumption_Lag_30"] = data["Daily_Consumption_kWh"].shift(30)
```

```python
data[[
    "Date",
    "Daily_Consumption_kWh",
    "Consumption_Lag_1",
    "Consumption_Lag_7",
    "Consumption_Lag_30"
]].head(35)
```

```python
data["Rolling_Mean_7"] = (
    data["Daily_Consumption_kWh"]
    .shift(1)
    .rolling(window=7)
    .mean()
)
```

```python
data["Rolling_Mean_30"] = (
    data["Daily_Consumption_kWh"]
    .shift(1)
    .rolling(window=30)
    .mean()
)
```

```python
data["Month_Sin"] = np.sin(2 * np.pi * data["Month"] / 12)

data["Month_Cos"] = np.cos(2 * np.pi * data["Month"] / 12)
```

```python
data.head(40)
```

```python
data = data.dropna().reset_index(drop=True)
```

```python
print("Missing values:")
print(data.isnull().sum())
```

```python
target = "Daily_Consumption_kWh"
```

```python
features = [
    "Year",
    "Month",
    "Day",
    "Day_of_Week",
    "Is_Weekend",
    "Consumption_Lag_1",
    "Consumption_Lag_7",
    "Consumption_Lag_30",
    "Rolling_Mean_7",
    "Rolling_Mean_30",
    "Month_Sin",
    "Month_Cos"
]
```

```python
X = data[features]

y = data[target]
```
